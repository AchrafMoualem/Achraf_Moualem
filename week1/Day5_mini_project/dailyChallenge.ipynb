# Sorting

```python
# Step 1: Get input from the user
words_input = input('Enter words separated by commas: ')

# Step 2: Split the string into a list
words_list = words_input.split(',')

# Step 3: Sort the list alphabetically
words_list.sort()

# Step 4: Join the sorted list back into a string
result = ','.join(words_list)

# Step 5: Print the result
print(result)
```

# Longest Word

```python
# Step 1: Define the function
def longest_word(sentence):

    # Step 2: Split the sentence into words
    words = sentence.split()

    # Step 3: Initialize variables
    longest = ''

    # Step 4 & 5: Iterate and compare lengths
    for word in words:
        if len(word) > len(longest):
            longest = word

    # Step 6: Return the longest word
    return longest


# Tests
print(longest_word("Margaret's toy is a pretty doll."))        # Margaret's
print(longest_word("A thing of beauty is a joy forever."))      # forever.
print(longest_word("Forgetfulness is by all means powerless!")) # Forgetfulness
```
