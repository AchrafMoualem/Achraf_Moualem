# Daily Challenge: Classification with Neural Networks in TensorFlow

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.datasets import make_circles
```

```python
from sklearn.datasets import make_circles

samples = 1000
X, y = make_circles(samples,
                    noise = 0.03,
                    random_state = 42)

print('X : ', X)
print('\n')
print('y : ', y)
```

```python
plt.figure(figsize=(6, 6))
plt.scatter(X[y == 0, 0], X[y == 0, 1], c="blue", label="Class 0", alpha=0.6)
plt.scatter(X[y == 1, 0], X[y == 1, 1], c="red",  label="Class 1", alpha=0.6)
plt.title("make_circles dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(4, activation=tf.nn.relu),
    tf.keras.layers.Dense(1, activation='sigmoid'),
])
```

```python
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"]
)
```

```python
model.fit(X, y, epochs=100)
accuracy = model.evaluate(X, y)[1]
accuracy
```

```python
model_2 = tf.keras.Sequential([
    tf.keras.layers.Dense(4, activation=tf.nn.relu),
    tf.keras.layers.Dense(1, activation='sigmoid'),
])
```

```python
model_2.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"]
)
model_2.fit(X, y, epochs=150)
accuracy_2 = model_2.evaluate(X, y)[1]
accuracy_2
```

```python
def plot_decision_boundary(model, X, y, title="Decision Boundary"):
    xx, yy = np.meshgrid(np.linspace(X[:,0].min()-0.1, X[:,0].max()+0.1, 200),
                         np.linspace(X[:,1].min()-0.1, X[:,1].max()+0.1, 200))

    preds = model.predict(np.c_[xx.ravel(), yy.ravel()], verbose=0).reshape(xx.shape)

    plt.figure(figsize=(6, 5))
    plt.contourf(xx, yy, preds, levels=50, cmap="RdBu", alpha=0.7)
    plt.contour(xx, yy, preds, levels=[0.5], colors="white", linewidths=1.5)
    plt.scatter(X[y==0,0], X[y==0,1], c="blue", label="class 0", alpha=0.6, s=15)
    plt.scatter(X[y==1,0], X[y==1,1], c="red",  label="class 1", alpha=0.6, s=15)
    plt.title(title)
    plt.legend()
    plt.show()
```

```python
plot_decision_boundary(model, X, y, title=f"Model 1 Decision Boundary (Acc: {accuracy:.2f})")
```

```python
plot_decision_boundary(model_2, X, y, title=f"Model 2 Decision Boundary (Acc: {accuracy:.2f})")
```
