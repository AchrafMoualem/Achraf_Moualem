# Daily Challenge: Classifying Handwritten Digits with CNNs

```python
import tensorflow as tf
import tensorflow.keras as keras 
```

```python
data = keras.datasets.mnist.load_data()
data[0][0].shape
```

```python
# Flatten the images from 28x28 to 784 pixels.
(x_train, y_train), (x_test, y_test) = data
x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)
```

```python
x_train.shape, x_test.shape
```

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
```

```python
x_train.std()
```

```python
# One-hot encode the target labels using keras.utils.to_categorical().
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)
```

```python
# Create a Sequential model.
model = keras.Sequential(
    [
        keras.layers.Input(shape=(784,)),
        keras.layers.Dense(128, activation="relu"),
        keras.layers.Dense(10, activation="softmax"),
    ]
)
```

```python
model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"]
)
model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
accuracy = model.evaluate(x_test, y_test)[1]
```

```python
accuracy
```

```python
# Reshape the input data to the shape expected by a Conv2D layer (e.g., (60000, 28, 28, 1)).
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)
```

```python
# cnn model
cnn = keras.Sequential(
    [
        keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
        keras.layers.MaxPooling2D((2, 2)),
        keras.layers.Conv2D(64, (3, 3), activation='relu'),
        keras.layers.MaxPooling2D((2, 2)),
        keras.layers.Flatten(),
        keras.layers.Dense(64, activation='relu'),
        keras.layers.Dense(10, activation='softmax')
    ]
)
cnn.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
cnn.fit(x_train, y_train, epochs=5, batch_size=32, validation_split
=0.2)
cnn_accuracy = cnn.evaluate(x_test, y_test)[1]
cnn_accuracy
```

as we can see the accuracy of the CNN model (99%) is bigger than the accuracy of the fully connected sequential model (97%), so our CNN model is more accurate as it capture more complex pattern and details.
